```python
#Lettura file dati mergiati ch4-bufale e robot, selezionando solo colonne che ci servono per questo momento.

mandria_path = './dati_bufale/dati_grezzi_agosto_luglio/Dataset_CH4/Bufale_Rinna/Luglio/protocollo_2/ch4_cutted_milkings.csv'
mandria = pd.read_csv(mandria_path, usecols = ['CH4', 'ID_animal', 'Milking_number', 'Produzione (kg)'])

mandria['ID_animal'] = mandria['ID_animal'].apply(lambda x: int(x))
mandria['Milking_number'] = mandria['Milking_number'].apply(lambda x: int(x))

animals = list(set(mandria['ID_animal']))
mungiture = list(set(mandria['Milking_number']))


print(len(animals))
print(len(mungiture))
```

```python
# Group by 'Milking_number' and calculate the mean of 'Methane'
average_methane_per_milking = mandria.groupby('Milking_number').mean().reset_index()

average_methane_per_milking.head()

#METANO MEDIO EMESSO PER MUNGITURA
mean_ch4_per_milking = average_methane_per_milking['CH4'].mean()
```

```python
# Group by 'Animal_ID' and count the unique 'Milking_number'
milking_count_per_animal = mandria.groupby('ID_animal')['Milking_number'].nunique().reset_index()

# Step 2: Group by 'Animal_ID' to calculate the average and standard deviation of the means
mean_and_std = average_methane_per_milking.groupby('ID_animal').agg({
    'CH4': ['mean', 'std'],
    'Produzione (kg)': 'mean'}).reset_index()

# Flatten the MultiIndex columns resulting from the aggregation
mean_and_std.columns = ['ID_animal', 'CH4_mean', 'CH4_std', 'avg_produzione']



```

```python

# Salva il DataFrame combinato in un nuovo file CSV.
output_file = './dati_bufale/dati_grezzi_agosto_luglio/Dataset_CH4/Bufale_Rinna/Luglio/protocollo_2/mean_ch4_per_animal.xlsx'
mean_and_std.to_excel(output_file, index=False)


output_file_2 = './dati_bufale/dati_grezzi_agosto_luglio/Dataset_CH4/Bufale_Rinna/Luglio/protocollo_2/mean_ch4_per_milking.xlsx'
average_methane_per_milking.to_excel(output_file_2, index=False)
print("File CSV combinato creato con successo.")

```
