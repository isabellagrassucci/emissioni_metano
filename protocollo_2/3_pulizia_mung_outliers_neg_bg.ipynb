```python
#Lettura file dati mergiati ch4-bufale e robot, selezionando solo colonne che ci servono per questo momento.

mandria_path = './dati_bufale/dati_grezzi_agosto_luglio/Luglio/protocollo_2/ch4_merged_sniffer_AMS_01-09_07_24.csv'
mandria = pd.read_csv(mandria_path, usecols = ['timestamp', 'CH4', 'Ora inizio', 'ID_animal', 'Milking_duration', 'Milking_number', 'Produzione (kg)', 'date'])
```

```python
mandria.tail()
```

```python
mandria['ID_animal'] = mandria['ID_animal'].apply(lambda x: int(x))
mandria['Milking_number'] = mandria['Milking_number'].apply(lambda x: int(x))
```

```python
#dividing and plotting each milking (keeping animal information)
#inspection of milking with calculus of mean, median, standard deviation without removal of data

animals = list(set(mandria['ID_animal']))
mungiture = list(set(mandria['Milking_number']))
print(len(animals))
print(len(mungiture))
```

```python
#Solution for first and first three minutes (to solve after)

mandria['progressive_seconds'] = mandria.groupby(['ID_animal', 'Milking_number']).cumcount() + 1
```

```python
matrix=[]

for mungitura in mungiture[0:100]:
    milking_ch4 = mandria[mandria['Milking_number'] == mungitura]
    milking_ch4 = milking_ch4.reset_index(drop=True)
    
    print(mungitura)
    animal = milking_ch4.iloc[0,5]
    date = milking_ch4.iloc[0,2]

    
    #milking_ch4 = milking_ch4[milking_ch4['progressive_seconds'] < 500]
    milking_ch4 = milking_ch4.set_index(milking_ch4['progressive_seconds'])
    
    #print(set(milking_ch4['Milking_number']))
    print(milking_ch4.shape)
    
    #print(milking_ch4.head())
    #print(milking_ch4.tail())

    
    #Plot singole mungiture 
    plt.figure(facecolor='white')
    milking_ch4.plot(x='progressive_seconds', y='CH4', linewidth=0.5, color = "blue")
    #plt.xlabel('Time')
    plt.ylabel('ppm')

    title = str(mungitura) + "-" + str(animal) + "-" + str(date) + "-original Luglio" 
    plt.title(title)
    
    '''
    directory = os.path.join("./dati_bufale/dati_grezzi_agosto_luglio/Luglio/protocollo_2/plot_bufale/", str(animal))
    if not os.path.exists(directory):
        os.makedirs(directory)
    file_path = os.path.join(directory, (str(mungitura) + "-original.png"))
    plt.savefig(file_path, format="png", dpi=250)
    '''
    
    plt.show()
    
    
    
    
    ch4_mean = milking_ch4['CH4'].mean()
    #milking_median = ch4['CH4'].median()
    #print("Milking median is: ", milking_median)
    std = np.std(milking_ch4['CH4'])
    #print('Milking std. deviation is', round(std, 2)) 
    threshold = round(round(std, 2)*3,2)
    #print('treshold is std x3: ', threshold)
     
    positive_outliers = round(ch4_mean + threshold, 2)
    negative_outliers = round(ch4_mean - threshold, 2)
    #print(positive_outliers, negative_outliers)    
    
    #print(ch4.shape)
    filtered_milking_ch4 = milking_ch4[milking_ch4['CH4'] < positive_outliers]
    filtered_milking_ch4 = filtered_milking_ch4[filtered_milking_ch4['CH4'] > negative_outliers]
        
    ch4_mean_out=filtered_milking_ch4['CH4'].mean()
    
    
    #Plot singole mungiture filtrate per outliers e background
    #plt.figure(facecolor='white')
    #plt.xlabel('Time')
    #plt.ylabel('ppm')
    #title = "CH4 " + str(mungitura) + "-" + str(cow) + "-no_out" 
    #plt.title(title)
    #plt.plot(filtered_milking_ch4['CH4'], linewidth=0.5, color = "red")
#
    #
    #directory = "C:/Users/chiar/Dropbox/Prove_MOOLOGGER/Troscia_frisone/Dati_giu_lug_2024/plot_mung_v3_no_neg/ch4_no_out/"
    #if not os.path.exists(directory):
    #    os.makedirs(directory)
    #file_path = directory + str(mungitura) + ".png"
    #plt.savefig(file_path, format="png", dpi=250)
    #
    #plt.show()
    
    

    ###RIMUOVI NEGATIVI
    #remove data below zero
    ch4_no_neg = filtered_milking_ch4[filtered_milking_ch4['CH4'] > 0]
        
    ch4_no_neg_mean = ch4_no_neg['CH4'].mean()
    
    
    ##Plot singole mungiture filtrate per outliers e background
    #plt.figure(facecolor='white')
    #plt.xlabel('Time')
    #plt.ylabel('ppm')
    #title = "CH4 " + str(mungitura) + "-" + str(cow) + "-no_out_no_neg" 
    #plt.title(title)
    #plt.plot(ch4_no_neg['CH4'], linewidth=0.5, color = "blue")
    #
    #directory = "C:/Users/chiar/Dropbox/Prove_MOOLOGGER/Troscia_frisone/Dati_giu_lug_2024/plot_mung_v3_no_neg/ch4_no_out_no_neg/"
    #if not os.path.exists(directory):
    #    os.makedirs(directory)
    #file_path = directory + str(mungitura) + ".png"
    #plt.savefig(file_path, format="png", dpi=250)
    #
    #plt.show()
    
    
    
    ######RIMOZIONE BACKGROUND PER MUNGITURE!!!!!!! 10 PUNTI PIU BASSI ***POSITIVI***

    #Calculus of background of each milking with minimum of 10 points
    #ch4 = pd.DataFrame(milking_ch4['CH4'])
   
    minim = list()    
    ch4 = ch4_no_neg
    #print(ch4.shape)
    for time in range(10):
        minimum = ch4['CH4'].min()
        minim.append(minimum)
        ch4 = ch4[ch4['CH4'] != minimum] #this way we cut that 10 data, but do not remove from all the data
       
     #print(ch4.shape)
    
    minim = pd.Series(minim)
    background = minim.mean()
    #print ("Average minimum background is: ", background)
    
    
    ch4_no_back = ch4_no_neg
    ch4_no_back['CH4'] = ch4_no_back['CH4'].apply(lambda x: x - background)

    
    #check that the dataframe with which we calculate the minimum is smaller than the original
    #and most important that the original and the no_back dataframe contains all the data, including all the minimums
    
    #print(ch4.shape)
    #print(milking_ch4.shape)
    #print(ch4_no_back.shape)
    
    no_back_mean = ch4_no_back['CH4'].mean()
    
    
    #plt.figure(facecolor='white')
    #plt.xlabel('Time')
    #plt.ylabel('ppm')
    #title = "CH4 " + str(mungitura) + "-" + str(cow) + "-no_out_no_neg_no_back" 
    #plt.title(title)
    #plt.plot(ch4_no_back['CH4'], linewidth=0.5, color = "black")
    
    
    #print(ch4.shape)
    
    
    #remove data below zero
    ch4_final = ch4_no_back[ch4_no_back['CH4'] > 0]
    
    ch4_final_mean = ch4_final['CH4'].mean()
    print("Final Average after removals: ", ch4_final_mean)
    
    #print(filtered_milking_ch4.shape)
    
    
    #Plot singole mungiture filtrate per outliers e background
    plt.figure(facecolor='white')
    ch4_final.plot(x='progressive_seconds', y='CH4', linewidth=0.5, color = "red")
    #plt.xlabel('Time')
    plt.ylabel('ppm')
        
    title = str(mungitura) + "-" + str(animal) + "-" + str(date) + "-cleaned Luglio" 
    plt.title(title)
    
    
    # Add a note (annotation)
    note_text = "media ppm= " + str(ch4_final_mean) 
    # Calculate the center of the plot
    x_center = (plt.xlim()[0] + plt.xlim()[1]) / 2
    y_below_axis = plt.ylim()[0] - (plt.ylim()[1] - plt.ylim()[0]) * 0.3  # 30% below the y-axis
    plt.text(x_center, y_below_axis, note_text, fontsize=12, ha='center', va='center', bbox=dict(facecolor='lightyellow', edgecolor='black'))
    
      
    # Adjust the plot to make sure the text is visible
    plt.subplots_adjust(bottom=0.3)  # Increase the bottom margin to ensure the annotation is included
    
    '''
    directory = os.path.join("./dati_bufale/dati_grezzi_luglio_agosto/protocollo_2/plot_bufale/", str(animal))
    if not os.path.exists(directory):
        os.makedirs(directory)
    file_path = os.path.join(directory, (str(mungitura) + "-cleaned" + ".png"))
    plt.savefig(file_path, format="png", dpi=250)
    '''
    
    
    plt.show()
    print(len(ch4_final['progressive_seconds']))
    
    
    matrix.append(ch4_final)
    
    
```
